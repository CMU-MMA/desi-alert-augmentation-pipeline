# Generate unit test data

This notebook (re)generates unit test data for the `desi_aap` package. It needs to be run manually.

```python
from upath import UPath

import lsdb

DATA_DIR = UPath("tests/desi_aap/data")
```

### DESI DR1 (bridges2)

Carve a small slice of the **COSMOS** field out of the DESI DR1 catalog.

[COSMOS](https://cosmos.astro.caltech.edu/page/astronomers) is a 2 square degree equatorial field centered at (150.11917, +2.20583).

```python
lsdb.read_hats(
    "/ocean/projects/phy250012p/shared/3DTS/DESI/dr1/desi_dr1_zcat",
).cone_search(
    ra=150.11917,
    dec=2.20583,
    radius_arcsec=600,
).write_catalog(
    DATA_DIR / "desi_dr1_cosmos",
    catalog_name="desi_dr1_cosmos",
    overwrite=True,
)
```
